```python
import numpy as np
```

```python
## create array using numpy
## create a 1D array
arr1 = np.array([1, 2, 3, 4, 5])
print(arr1)
print(type(arr1))
print(arr1.shape)  # shape of the array
```

```python
arr2 = np.array([1, 2, 3, 4, 5])
arr2.reshape(1, 5)  # reshape the array to 1 row and 5 columns
#print(arr2)
```

```python
print(arr2.shape)  # shape of the array
```

```python
##  Attributes of Numpy
arr = np.array([[1, 2, 3], [4, 5, 6]])

print("Array:\n", arr)
print("Shape:", arr.shape)  # Output: (2, 3)
print("Number of dimensions:", arr.ndim)  # Output: 2
print("Size (number of elements):", arr.size)  # Output: 6
print("Data type:", arr.dtype)  # Output: int32(may vary based on platform and version)
print("Item size (in bytes):", arr.itemsize)  # Output: 8 (may vary based on platform and version)
```

```python
### Numpy Vectorized Operation
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10,20,30,40,50])

### Element Wise Addition
print("Addition:", arr1 + arr2)  

### Element Wise Subtraction
print("Subtraction:", arr1 - arr2)  

### Element Wise Multiplication
print("Multiplication:", arr1 * arr2)

### Element Wise Division
```

### Universal Functions - (UFuncs)

```python
## Universal Functions
arr = np.array([2,3,4,5,6])

## square root
print(np.sqrt(arr))  

## Exponential
print(np.exp(arr))  

## Sine 
print(np.sin(arr))

## natural log
print(np.log(arr))
```

```python
## DataFrame
## Create a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)
print("DataFrame:\n", df)
print(type(df))
```

```python
## Create a DataFrame from a list of Dictionaries
data = [
    {'Name': 'Alice', 'Age': 25, 'City': 'New York'},
    {'Name': 'Bob', 'Age': 30, 'City': 'Los Angeles'},
    {'Name': 'Charlie', 'Age': 35, 'City': 'Chicago'}
]
df_list = pd.DataFrame(data)
print("DataFrame from list of dictionaries:\n", df_list)
print(type(df_list))
```

```python
## Adding a column with a default value
df['Salary'] = [1000] * len(df)  # Assigning a default value of 1000 to all rows
df
```

```python
## Remove a column 
df.drop('Salary', axis=1, inplace=True)  # Remove the 'Salary' column
```

```python
df
```

```python
## Add +1 to all values in the 'Sales' column
df['Sales'] = df['Sales'] + 1
df
```

```python
df.describe()  # statistical summary of the DataFrame
```

```python
# Handling missing values
df.isnull().any(axis=1)  # rows with missing values
```

```python
df.isnull().sum()  # count of missing values in each column
```

```python
#filling missing values with the mean of the column
df['Sales_fillNA'] = df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df.dtypes
```

```python
## Renaming Columns
df = df.rename(columns ={'Date':'Sales Date'})
```

```python
df.head()
```

```python
## change datatypes
df['Value'] = df['Value'].fillna(df['Value'].mean())
df['Value_new'] = df['Value'].astype('int')
df.head()
```

```python
df['New Value'] = df['Value'].apply(lambda x: x * 2)
df.head()
```

```python
## Data Aggregation and Grouping
grouped_mean= df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['Product', 'Region'])['Value'].sum()
print(grouped_sum)
```

```python
## aggreagte with multiple functions
grouped_agg = df.groupby('Region')['Value'].agg(['mean', 'sum', 'count'])
print(grouped_agg)
```

```python
## Merging and joining Dataframes
# Create a sample DataFrame
df1 = pd.DataFrame({
    'Key' : ['A', 'B', 'C'],
    'Value1' : [1, 2, 3]
})
df2 = pd.DataFrame({
    'Key' : ['A', 'B', 'D'],
    'Value2' : [4, 5, 6]
})
```

```python
df1
```

```python
df2
```

```python
## Merge Dateframe on the 'Key Columns'
pd.merge(df1, df2, on='Key', how='inner')  # Inner join
```

```python
pd.merge(df1, df2, on='Key', how='outer')  # out join
```

```python
pd.merge(df1, df2, on='Key', how='left')  # left join
```
